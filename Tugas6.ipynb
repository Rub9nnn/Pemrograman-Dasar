```python
def hitung_rata_rata(data_nilai):
  total = 0
# Perulangan untuk menjumlahkan hasil
  for nilai in data_nilai:
    total += nilai
  # Pembagian untuk rata-rata
  if total > 0:
    return total/ len(data_nilai)
  return 0

def hitung_rata_rata_modifikasi(data_nilai):

  total_nilai_valid = 0
  jumlah_nilai_valid = 0

  for nilai in data_nilai:
    if 0 <= nilai <= 100:
      total_nilai_valid += nilai
      jumlah_nilai_valid += 1
    else:
        print(f"Nilai {nilai} diabaikan karena diluar rentang (0-100).")

  if jumlah_nilai_valid > 0:
    return total_nilai_valid / jumlah_nilai_valid
  return 0

  nilai_praktikan = [85, 78, 92, 65, 90, 105, -5]
  rata_rata = hitung_rata_rata_modifikasi(nilai_praktikan)

  print("n/==================================")
  print(f"List Nilai: {nilai_praktikan}")
  print(f"Nilai rata-rata kelas (setelah validasi): {rata_rata}")

```

```python
def tentukan_predikat_modifikasi(skor):

  if not isinstance(skor, (int, float)):
    return "ERROR: Input harus berupa angka (int/float)."

  if skor >= 85:
    return "A"
  elif skor >= 70:
    return "B"
  elif skor >= 60:
    return "C"
  else:
    return "E"

print("=============================")
print(f"Skor 75 mendapat predikat: {tentukan_predikat_modifikasi(75)}")
print(f"Skor 85.5 mendapatkan predikat: {tentukan_predikat_modifikasi(85.5)}")
print(f"Skor 'ABC' mendapatkan predikat: {tentukan_predikat_modifikasi('ABC')}")
print("================================")
```

```python

def hitung_faktorial_modifikasi(n):
  if n== 0:
    return 1
  elif n < 0:
    return "ERROR: Faktorial tidak terdefenisi untuk bilangan negatif."
  elif not isinstance(n, int):
    return "ERROR: Input harus bilangan bulat."
  hasil = 1
  while n > 1:
    hasil *= n
    n -= 1

  return hasil

print("n/======================================")
print(f"Faktorial dari 5 adalah: {hitung_faktorial_modifikasi(5)}")
print(f"Faktorial dari 0 adalah: {hitung_faktorial_modifikasi(0)}")
print(f"Faktorial dari -4 adalah: { hitung_faktorial_modifikasi(-4)}")
print("========================================================")
```

```python
def analisis_suhu_modifikasi(data_suhu, batas_panas=35, batas_dingin=15):
  panas_count = 0
  dingin_count = 0

  for suhu in data_suhu:
    if suhu >= batas_panas:
      panas_count += 1
    elif suhu <= batas_dingin:
      dingin_count += 1

  return panas_count, dingin_count

suhu_mingguan = [36, 30, 14, 25, 38, 16, 28]
p1, d1 = analisis_suhu_modifikasi(suhu_mingguan)
print("=======================================")
print("Analisis default (panas >=35, Dingin<=15):")
print(f"Jumlah hari panas: {p1}")
print(f"Junlah hari dingin: {d1}")

p2, d2 = analisis_suhu_modifikasi(suhu_mingguan, batas_panas=30, batas_dingin=20)
print("n/Analisis kustom (Panas>=30, Dingin<=20):")
print(f"Jumlah hari panas: {p2}")
print(f"Jumlah hari dingin: {d2}")
print("===================================")
```
