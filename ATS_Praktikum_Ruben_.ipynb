```python
kapasitas_maks = float(input("Masukkan kapasitas maksimum (W): "))
beban_terpakai = float(input("Masukkan beban terpakai ( W): "))

presentase = (beban_terpakai / kapasitas_maks) * 100
print(f"Presentase pemakaian: {presentase:2f}%")

if presentase <= 60:
  status = "Beban ringan"
elif  presentase <= 85:
  status = "Beban sedang"
else:
  status = "Beban berat"
print(f"Status beban listrik: {status}")
```

```python
def hitung_efisiensni_termal(energi_keluar, energi_masuk):
 efisiensi = (energi_keluar / energi_masuk) * 100
 return efisiensi

energi_masuk = float(input("Masukkan energi (joule): "))
energi_keluar = float(input("Masukkan energi keluar (Joule): "))

hasil_efisiensi =  hitung_efisiensni_termal(energi_keluar, energi_masuk)

print(f"Efisiensi termal turbin gas: {hasil_efisiensi:.2f}%")
```
